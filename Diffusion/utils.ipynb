```python
import json

with open('/workspace/annotations/captions_train2017.json', 'r') as f:
    datas = json.load(f)

from tqdm import tqdm

images = datas['images']
annots = datas['annotations']

def find_from_images(id):
    for i in images:
        if i['id'] == id:
            return i['file_name']

val_data = []
ids = set([])
for anot in tqdm(annots):
    iid = anot['image_id']
    if iid not in ids:
        ids.add(iid)
        file_name = find_from_images(iid)
        val_data.append({
            'caption': anot['caption'],
            'file_name': file_name
        })

# JSON 파일로 저장
with open('./train.json', 'w') as file:
    json.dump(val_data, file, indent=4)  # indent=4는 보기 좋게 들여쓰기
```
